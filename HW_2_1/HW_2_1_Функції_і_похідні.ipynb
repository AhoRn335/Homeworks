1. Задана функція залежності оцінки від кількості годин, витрачених на навчання: $ f(x) = 10x - 5 $.
  1. З допомогою Python обчисліть похідну цієї функції.
  2. Використовуйте бібліотеку `matplotlib` для побудови графіків функції та її похідної в діапазоні значень від 0 до 10 годин. Проаналізуйте, як зміна кількості годин вивчення впливає на оцінку.



```python
import matplotlib.pyplot as plt
import numpy as np
import sympy as sp
```

```python
def sympy_df(function):
    x = sp.symbols('x')

    f = function(x)
    df = sp.diff(f, x)
    return df
```

```python
def func_plot(data, function, derivative, simple_function=False):
    f = function(data)
    df = derivative(data)
    
    x = sp.symbols('x')
    plt.figure(figsize=(10, 5))
    if simple_function:
        plt.plot(data, f, label=f'f(x) = {function.__name__}', color='blue')
        plt.plot(data, df, label=f"f'(x) = {derivative.__name__}", color='green')
    else:
        plt.plot(data, f, label=f'f(x) = {function.__name__}(x)', color='blue')
        plt.plot(data, df, label=f"f'(x) = {derivative.__name__}(x)", color='green')

    plt.title("Графік функції та її похідної")
    plt.xlabel("x")
    plt.ylabel("y")
    plt.legend()
    plt.grid(True)

    plt.show()
```

A. З допомогою Python обчисліть похідну цієї функції.

```python
def lin_func(x):
    return 10*x - 5
lin_func.__name__ = '10*x - 5'

def d_lin_func(x):
    return np.full_like(x, 10)
d_lin_func.__name__ = '10'
```

```python
sympy_df(lin_func)
```

2. Використовуйте бібліотеку `matplotlib` для побудови графіків функції та її похідної в діапазоні значень від 0 до 10 годин. Проаналізуйте, як зміна кількості годин вивчення впливає на оцінку.

```python
xs = np.linspace(0,10,10)
func_plot(xs, lin_func, d_lin_func, simple_function=True)
```

Так як функція, яка описує залежність оцінки від кількості витрачених на навчання годин, є лінійною, то зі збільшенням кількості годин (аргументу) відбувається збільшення значення оцінки (залежної величини). Тому можемо стверджувати, що збільшення кількості навчальних годин передбачає підвищення оцінки.

```python
sympy_df(sp.sin)
```
